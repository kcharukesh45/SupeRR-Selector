```python

import pandas as pd

balls = pd.read_csv(r"C:\Users\Admin\Downloads\New folder\ball_by_ball.csv")

# Balls faced: wides are not counted
balls["balls_faced"] = balls["extras_type"] != "wide"

# Count only the batter who was dismissed
balls["batter_out"] = (
    (balls["is_wicket"] == 1) &
    (balls["dismissed_player_id"] == balls["striker_id"])
)

batting = balls.groupby(["striker_id", "striker_name"]).agg(
    runs=("runs", "sum"),
    balls=("balls_faced", "sum"),
    fours=("is_four", "sum"),
    sixes=("is_six", "sum"),
    dismissals=("batter_out", "sum"),
    matches=("match_id", "nunique")
).reset_index()

batting["strike_rate"] = (
    batting["runs"] / batting["balls"] * 100
).round(2)

batting = batting.sort_values("runs", ascending=False)

batting.to_csv("Results/batting_stats.csv", index=False)

print("===== TOP 15 BATTERS BY RUNS =====")
print(batting.head(15).to_string(index=False))
```
