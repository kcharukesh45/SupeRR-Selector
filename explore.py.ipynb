```python
import pandas as pd

# Load datasets
balls = pd.read_csv(r"C:\Users\Admin\Downloads\New folder\ball_by_ball.csv")
matches = pd.read_csv(r"C:\Users\Admin\Downloads\New folder\matches.csv")

# 1. Check innings
print("\n===== INNINGS =====")
print(balls["innings"].value_counts().sort_index())

# 2. Check overs
print("\n===== OVER INFORMATION =====")
print("Minimum over:", balls["overid"].min())
print("Maximum over:", balls["overid"].max())

# 3. Check runs
print("\n===== RUNS =====")
print("Total runs:", balls["runs"].sum())
print("Total fours:", balls["is_four"].sum())
print("Total sixes:", balls["is_six"].sum())

# 4. Check wickets
print("\n===== WICKETS =====")
print("Total wickets:", balls["is_wicket"].sum())

# 5. Check match results
print("\n===== MATCH RESULTS =====")
print(matches["result"].value_counts().head(10))

# 6. Check player performances
print("\n===== TOP RUN SCORERS =====")
batting = balls.groupby("striker_name")["runs"].sum()
print(batting.sort_values(ascending=False).head(10))

print("\nEXPLORATION COMPLETED!")
```
