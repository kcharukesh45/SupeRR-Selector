```python
import pandas as pd

balls = pd.read_csv(r"C:\Users\Admin\Downloads\New folder\ball_by_ball.csv")

# Keep normal innings
balls = balls[balls["innings"].isin([1, 2])].copy()

# Check total deliveries per innings
check = balls.groupby(
    ["match_id", "innings"]
).agg(
    total_deliveries=("match_id", "size"),
    extras=("extras_type", lambda x: x.notna().sum()),
    max_over=("overid", "max")
).reset_index()

# Show innings with more than 120 deliveries
print("\n===== INNINGS WITH MORE THAN 120 BALLS =====")
print(check[check["total_deliveries"] > 120].head(20))

# Check extra types
print("\n===== EXTRA TYPES =====")
print(balls["extras_type"].value_counts(dropna=False))

# Check deliveries in the 20th over
print("\n===== 20TH OVER EXAMPLES =====")
print(balls[balls["overid"] == 20][
    ["match_id", "innings", "overid",
     "over_ball_num", "extras_type"]
].head(20))
```
