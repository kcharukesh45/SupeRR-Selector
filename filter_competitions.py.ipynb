```python

import pandas as pd
import os

balls = pd.read_csv(r"C:\Users\Admin\Downloads\New folder\ball_by_ball.csv")
matches = pd.read_csv(r"C:\Users\Admin\Downloads\New folder\matches.csv")
comps = pd.read_csv(r"C:\Users\Admin\Downloads\New folder\competitions.csv")

# Select IPL and T20I competitions
selected = comps[
    comps["comp_name"].str.contains("IPL", case=False, na=False) |
    (comps["match_format"] == "T20I")
]

# Get matching match IDs
matches = matches[matches["comp_id"].isin(selected["comp_id"])]

# Keep only deliveries from selected matches
balls = balls[balls["match_id"].isin(matches["match_id"])]

os.makedirs("Results", exist_ok=True)

balls.to_csv("Results/ipl_t20i_balls.csv", index=False)
matches.to_csv("Results/ipl_t20i_matches.csv", index=False)

print("Selected matches:", len(matches))
print("Selected deliveries:", len(balls))
print("Saved successfully!")
```
