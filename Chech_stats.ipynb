```python
import pandas as pd

balls = pd.read_csv(r"C:\Users\Admin\Downloads\New folder\ball_by_ball.csv")

# Total runs
print("Total runs:", balls["runs"].sum())

# Extras
print("\n===== EXTRAS =====")
print(balls["extras_type"].value_counts())

print("\nExtras 1:", balls["extras1_runs"].sum())
print("Extras 2:", balls["extras_2_runs"].sum())

# Wicket information
print("\n===== WICKET TYPES =====")
print(balls[balls["is_wicket"] == True]
      ["dismissal_type"].value_counts())

# Sample extra deliveries
print("\n===== EXTRA DELIVERY EXAMPLES =====")
print(balls[balls["extras_type"].notna()]
      [["striker_name", "bowler_name", "runs",
        "extras_type", "extras1_runs",
        "extras_2_runs"]].head(10))

# Check innings 3 and 4
print("\n===== SUPER OVER CHECK =====")
print(balls[balls["innings"] >= 3]
      [["match_id", "innings", "overid", "runs"]]
      .head(10))
```
