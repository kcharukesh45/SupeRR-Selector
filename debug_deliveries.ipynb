```python
import pandas as pd

balls = pd.read_csv(r"C:\Users\Admin\Downloads\New folder\ball_by_ball.csv")

# Check one unusual innings
data = balls[
    (balls["match_id"] == 13513) &
    (balls["innings"] == 1)
].copy()

print("===== DELIVERY COUNT =====")
print("Total:", len(data))

print("\n===== EXTRA TYPES =====")
print(data["extras_type"].value_counts(dropna=False))

print("\n===== OVERSTEP DETAILS =====")
print(data[data["extras_type"] == "Overstep"][
    ["overid", "over_ball_num", "innings_ball_num",
     "runs", "extras1_runs", "extras_2_runs"]
].to_string(index=False))

print("\n===== WIDE DETAILS =====")
print(data[data["extras_type"] == "wide"][
    ["overid", "over_ball_num", "innings_ball_num",
     "runs", "extras1_runs"]
].to_string(index=False))

print("\n===== LEGAL BALL CALCULATION =====")
illegal = data["extras_type"].isin(["wide", "Overstep"])

print("Total deliveries:", len(data))
print("Illegal deliveries:", illegal.sum())
print("Legal deliveries:", (~illegal).sum())
```

```python
import pandas as pd

balls = pd.read_csv(r"C:\Users\Admin\Downloads\New folder\ball_by_ball.csv")

data = balls[
    (balls["match_id"] == 13513) &
    (balls["innings"] == 1)
].copy()

data["legal_ball"] = ~data["extras_type"].isin(["wide", "Overstep"])

check = data.groupby("overid").agg(
    total_deliveries=("match_id", "count"),
    legal_balls=("legal_ball", "sum"),
    extras=("extras_type", lambda x: x.notna().sum())
)

print(check)
```

```python
balls["legal_ball"] = ~balls["extras_type"].isin(
    ["wide", "Overstep"]
)

# Count deliveries and legal balls in each innings
check = balls.groupby(["match_id", "innings"]).agg(
    deliveries=("match_id", "count"),
    legal_balls=("legal_ball", "sum")
).reset_index()

print("Total innings checked:", len(check))

print("\nInnings with more than 120 legal balls:")
print(check[check["legal_balls"] > 120].to_string(index=False))

print("\nInnings with fewer than 120 legal balls:")
print(check[check["legal_balls"] < 120].head(15).to_string(index=False))

print("\nLegal ball summary:")
print(check["legal_balls"].describe())
```
